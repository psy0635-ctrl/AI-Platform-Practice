````python
# ============================================================
# [실습 1-2] 첫 화면 — 코드 네 줄
# ------------------------------------------------------------
# gr.Interface 는 함수 하나를 웹 화면으로 바꿔 주는 도구입니다.
#   fn      : 실행할 함수
#   inputs  : 입력칸의 종류
#   outputs : 출력칸의 종류
# 실행하면 이 셀 아래에 화면이 바로 나타납니다.
# ============================================================

def say_hello(name):
    """이름을 받아 인사말을 돌려준다"""
    return f"안녕하세요, {name}님!"


demo = gr.Interface(fn=say_hello, inputs="text", outputs="text")
demo.launch()
````

# STEP 2. Interface 의 세 가지 재료  (약 25분)

`gr.Interface` 에 넣는 것은 **함수 · 입력칸 · 출력칸** 세 가지입니다.
이 셋만 바꾸면 어떤 화면이든 만들 수 있습니다.

````python
# ============================================================
# [실습 2-1] 입력칸 종류 바꿔 보기
# ------------------------------------------------------------
# "text" 처럼 이름만 적는 간단한 방법도 되지만,
# gr.Textbox( ) 처럼 직접 쓰면 이름표·줄 수·설명을 정할 수 있습니다.
# ============================================================

def count_letters(text):
    """글자 수와 단어 수를 세어 돌려준다"""
    letter_count = len(text.replace(" ", ""))
    word_count = len(text.split())
    return f"공백 제외 {letter_count}자 · {word_count}단어"


counter_demo = gr.Interface(
    fn=count_letters,
    inputs=gr.Textbox(label="문장을 넣으세요", lines=3,
                      placeholder="여기에 입력"),
    outputs=gr.Textbox(label="세어 본 결과"))

counter_demo.launch()
````

````python
# ============================================================
# [실습 2-2] 입력이 두 개, 출력이 두 개
# ------------------------------------------------------------
# inputs 와 outputs 에 리스트를 넣으면 칸이 여러 개가 됩니다.
# 함수의 매개변수 순서와 inputs 순서가 짝이 맞아야 합니다.
#   Slider : 값을 끌어서 정하는 막대
# ============================================================

counter_demo.close()


def repeat_text(text, times):
    """문장을 여러 번 되풀이하고, 전체 길이도 함께 돌려준다"""
    repeated = (text + " ") * int(times)
    return repeated.strip(), f"전체 {len(repeated.strip())}자"


repeat_demo = gr.Interface(
    fn=repeat_text,
    inputs=[gr.Textbox(label="문장"),
            gr.Slider(minimum=1, maximum=10, value=3, step=1, label="몇 번")],
    outputs=[gr.Textbox(label="결과"),
             gr.Textbox(label="길이")])

repeat_demo.launch()
````

````python
# ============================================================
# [실습 2-3] 제목 · 설명 · 예시 넣기
# ------------------------------------------------------------
#   title       : 화면 맨 위 제목
#   description : 제목 아래 설명
#   examples    : 눌러서 바로 채워지는 예시 (학생이 무엇을 넣을지 몰라 멈추는 것을 막음)
# ============================================================

repeat_demo.close()


def make_id_card(name, department, year):
    """이름·학과·학년을 받아 한 줄 소개를 만든다"""
    return f"{department} {int(year)}학년 {name}"


card_demo = gr.Interface(
    fn=make_id_card,
    inputs=[gr.Textbox(label="이름"),
            gr.Textbox(label="학과"),
            gr.Slider(1, 4, value=1, step=1, label="학년")],
    outputs=gr.Textbox(label="소개"),
    title="한 줄 소개 만들기",
    description="이름과 학과를 넣으면 소개 문장을 만들어 줍니다.",
    examples=[["홍길동", "인공지능소프트웨어과", 1],
              ["김영희", "기계과", 2]])

card_demo.launch()
````

# STEP 3. 사진 분류기 화면  (약 25분)

5주차에 만든 `classifier` 를 그대로 가져와 화면에 붙입니다.
새로 배우는 것은 **`gr.Image` 와 `gr.Label` 두 컴포넌트**뿐입니다.

````python
# ============================================================
# [실습 3-1] 5주차 모델 불러오기
# ------------------------------------------------------------
# 5주차와 완전히 같은 코드입니다.
# 첫 실행은 모델(약 330MB)을 내려받느라 1~2분 걸립니다.
# ============================================================

card_demo.close()

from transformers import pipeline

classifier = pipeline("image-classification", model="google/vit-base-patch16-224")

print("모델 준비 완료 :", classifier.model.name_or_path)
print("구분하는 개수  :", len(classifier.model.config.id2label), "가지")
````

````python
# ============================================================
# [실습 3-2] 화면에 넣을 함수 만들기
# ------------------------------------------------------------
# ★ gr.Label 은 {이름표: 점수} 형태의 딕셔너리를 받습니다.
#   모델이 주는 [{"label":..., "score":...}, ...] 를 그 형태로 바꿔 줘야 합니다.
# ★ 입력이 비었을 때(None)를 반드시 처리합니다.
#   사진을 올리지 않고 Submit 을 누르는 학생이 꼭 있습니다.
# ============================================================

def classify_image(image):
    """사진을 받아 {이름표: 점수} 를 돌려준다"""
    if image is None:
        return {}
    results = classifier(image, top_k=5)
    return {item["label"]: float(item["score"]) for item in results}


# 화면을 만들기 전에 함수만 먼저 시험해 봅니다 (오류를 빨리 찾는 방법)
from PIL import Image
import requests

sample_url = ("https://huggingface.co/datasets/huggingface/"
              "documentation-images/resolve/main/coco_sample.png")
sample_image = Image.open(requests.get(sample_url, stream=True).raw).convert("RGB")
sample_image.save("sample.png")        # examples 에 쓰려고 파일로 저장

print(classify_image(sample_image))
````

````python
# ============================================================
# [실습 3-3] 사진 분류기 화면 완성
# ------------------------------------------------------------
#   gr.Image(type="pil")      : 올린 사진을 PIL 형식으로 함수에 넘긴다
#   gr.Label(num_top_classes) : 점수를 막대로 그려 준다
# type="pil" 을 빼면 numpy 배열이 넘어와서 모델이 받지 못합니다.
# ============================================================

image_demo = gr.Interface(
    fn=classify_image,
    inputs=gr.Image(type="pil", label="사진을 올리세요"),
    outputs=gr.Label(num_top_classes=5, label="분류 결과"),
    title="사진 분류기",
    description="사진을 올리면 무엇인지 맞혀 봅니다. ImageNet 1000가지 안에서 고릅니다.",
    examples=[["sample.png"]])

image_demo.launch()
````

> ### 💡 `type="pil"` 이 중요한 이유
>
> `gr.Image` 는 기본값으로 사진을 **numpy 배열**로 함수에 넘깁니다.
> transformers 모델은 PIL 이미지를 기대하므로, `type="pil"` 을 지정해야 그대로 받을 수 있습니다.
>
> | 지정 | 함수가 받는 것 |
> |---|---|
> | `gr.Image()` | numpy 배열 |
> | `gr.Image(type="pil")` | PIL 이미지 ← 오늘 쓰는 것 |
> | `gr.Image(type="filepath")` | 저장된 파일의 경로(문자열) |

# STEP 4. 객체 인식 화면 — 상자를 그려 주는 컴포넌트  (약 20분)

5주차에는 상자를 `ImageDraw` 로 직접 그렸습니다.
Gradio 에는 **`gr.AnnotatedImage`** 가 있어서 좌표만 넘기면 알아서 그려 줍니다.

````python
# ============================================================
# [실습 4-1] 객체 인식 함수 만들기
# ------------------------------------------------------------
# ★ gr.AnnotatedImage 는 (사진, 상자목록) 두 개짜리 묶음을 받습니다.
#   상자목록의 한 칸은 ((왼쪽, 위, 오른쪽, 아래), "이름표") 형태입니다.
#   5주차의 box 딕셔너리를 이 형태로 바꿔 주면 됩니다.
# ============================================================

image_demo.close()

detector = pipeline("object-detection", model="hustvl/yolos-tiny")


def detect_objects(image, threshold):
    """사진을 받아 (사진, 상자목록) 을 돌려준다"""
    if image is None:
        return None
    boxes = []
    for item in detector(image, threshold=threshold):
        box = item["box"]
        boxes.append(((box["xmin"], box["ymin"], box["xmax"], box["ymax"]),
                      f'{item["label"]} {item["score"]:.2f}'))
    return (image, boxes)


result = detect_objects(sample_image, 0.5)
print("찾은 물체 수 :", len(result[1]))
print("첫 번째 상자 :", result[1][0] if result[1] else "없음")
````

````python
# ============================================================
# [실습 4-2] 기준 점수를 화면에서 조절하기
# ------------------------------------------------------------
# Slider 를 붙이면 학생이 threshold 를 끌어서 바꿔 볼 수 있습니다.
# 5주차에는 코드를 고쳐야 했던 것을 화면에서 바로 바꾸는 셈입니다.
# ============================================================

detect_demo = gr.Interface(
    fn=detect_objects,
    inputs=[gr.Image(type="pil", label="사진"),
            gr.Slider(minimum=0.1, maximum=0.95, value=0.5, step=0.05,
                      label="기준 점수 (낮추면 많이 찾음)")],
    outputs=gr.AnnotatedImage(label="찾은 물체"),
    title="물체 찾기",
    description="사진 속 물체의 위치를 네모로 표시합니다. COCO 80가지를 찾습니다.",
    examples=[["sample.png", 0.5]])

detect_demo.launch()
````

> ### 여기서 확인할 것
>
> 막대를 왼쪽으로 끌면(0.1) 상자가 많아지고, 오른쪽으로 끌면(0.9) 확실한 것만 남습니다.
> **5주차에서 코드를 고쳐 가며 확인했던 것을 화면에서 바로 비교할 수 있습니다.**
> 이것이 데모를 만드는 이유입니다. 값을 바꿔 보는 사람이 코드를 몰라도 됩니다.

# STEP 5. 음성 받아쓰기 화면 — 마이크 버튼  (약 20분)

`gr.Audio` 는 **파일 업로드와 마이크 녹음을 한 컴포넌트로** 처리합니다.
5주차에서 파일을 올리고 형식을 바꾸던 과정이 한 줄로 줄어듭니다.

````python
# ============================================================
# [실습 5-1] Whisper 불러오기
# ------------------------------------------------------------
# 5주차와 같은 모델입니다. 약 1GB 라 2~3분 걸립니다.
# ============================================================

detect_demo.close()

transcriber = pipeline("automatic-speech-recognition", model="openai/whisper-small")

print("음성 모델 준비 완료")
````

````python
# ============================================================
# [실습 5-2] 받아쓰기 함수와 화면
# ------------------------------------------------------------
# ★ gr.Audio 옵션
#   sources=["upload", "microphone"] : 파일 업로드 칸과 마이크 버튼을 함께 보여 준다
#   type="filepath"                  : 녹음·업로드한 소리를 파일로 저장한 뒤 경로를 넘긴다
#                                      -> Whisper 가 경로를 그대로 받으므로 변환이 필요 없다
# ★ language="ko" 를 지정해야 한국어로 받아씁니다 (5주차에서 배운 함정)
# ============================================================

def transcribe_audio(audio_path):
    """음성 파일 경로를 받아 한국어로 받아쓴 글을 돌려준다"""
    if audio_path is None:
        return "소리를 올리거나 마이크로 녹음해 주세요."
    result = transcriber(audio_path,
                         chunk_length_s=30,
                         generate_kwargs={"language": "ko", "task": "transcribe"})
    return result["text"].strip()


audio_demo = gr.Interface(
    fn=transcribe_audio,
    inputs=gr.Audio(sources=["upload", "microphone"], type="filepath",
                    label="마이크를 누르고 말하거나, 파일을 올리세요"),
    outputs=gr.Textbox(label="받아쓴 글", lines=4),
    title="받아쓰기",
    description="한국어 음성을 글자로 바꿉니다. 20초 이내가 빠릅니다.")

audio_demo.launch()
````

# STEP 6. Blocks 로 화면 직접 배치하기  (약 25분)

`gr.Interface` 는 화면 배치를 알아서 정해 줍니다. 편하지만 바꿀 수 없습니다.
**`gr.Blocks`** 를 쓰면 어디에 무엇을 놓을지 직접 정할 수 있습니다.

````python
# ============================================================
# [실습 6-1] Blocks 기본 — Row 와 Column
# ------------------------------------------------------------
#   with gr.Row( )    : 안에 놓은 것들을 가로로 나란히
#   with gr.Column( ) : 안에 놓은 것들을 세로로 쌓기
#   button.click( )   : 버튼을 눌렀을 때 실행할 함수를 연결
# Interface 와 달리 "무엇을 눌렀을 때 무엇을 한다"를 직접 적어 줍니다.
# ============================================================

audio_demo.close()

with gr.Blocks(title="사진 분류기 (Blocks 버전)") as blocks_demo:
    gr.Markdown("## 사진 분류기\n사진을 올리고 버튼을 누르세요.")

    with gr.Row():
        with gr.Column():
            image_input = gr.Image(type="pil", label="사진")
            run_button = gr.Button("분류하기", variant="primary")
        with gr.Column():
            label_output = gr.Label(num_top_classes=5, label="결과")

    run_button.click(fn=classify_image, inputs=image_input, outputs=label_output)

blocks_demo.launch()
````

````python
# ============================================================
# [실습 6-2] 탭으로 세 데모를 하나로 합치기
# ------------------------------------------------------------
# gr.Tab 하나가 화면 위쪽의 탭 하나가 됩니다.
# 이미 만든 함수를 그대로 재사용하므로 새로 쓸 코드가 거의 없습니다.
# ============================================================

blocks_demo.close()

with gr.Blocks(title="AI 데모 모음") as all_in_one:
    gr.Markdown("# 인공지능소프트웨어과 AI 데모\n"
                "탭을 눌러 세 가지 기능을 사용해 보세요.")

    with gr.Tabs():
        with gr.Tab("사진 분류"):
            gr.Interface(fn=classify_image,
                         inputs=gr.Image(type="pil", label="사진"),
                         outputs=gr.Label(num_top_classes=5, label="결과"),
                         examples=[["sample.png"]])

        with gr.Tab("물체 찾기"):
            gr.Interface(fn=detect_objects,
                         inputs=[gr.Image(type="pil", label="사진"),
                                 gr.Slider(0.1, 0.95, value=0.5, step=0.05,
                                           label="기준 점수")],
                         outputs=gr.AnnotatedImage(label="찾은 물체"),
                         examples=[["sample.png", 0.5]])

        with gr.Tab("받아쓰기"):
            gr.Interface(fn=transcribe_audio,
                         inputs=gr.Audio(sources=["upload", "microphone"],
                                         type="filepath", label="음성"),
                         outputs=gr.Textbox(label="받아쓴 글", lines=4))

all_in_one.launch()
````
